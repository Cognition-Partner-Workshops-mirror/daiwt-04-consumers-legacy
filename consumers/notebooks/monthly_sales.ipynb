```python
import os
os.environ.setdefault("MSSQL_HOST", "daiwt-02-sqlserver.c0x4mes64o5e.us-east-1.rds.amazonaws.com")
import sys
from pathlib import Path
import pandas as pd
root = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / 'consumers/api/db.py').exists())
sys.path.insert(0, str(root))
from consumers.api.db import connect
from contextlib import closing
sql = """SELECT DATEPART(year,[Invoice Date Key]) AS sales_year, DATEPART(month,[Invoice Date Key]) AS sales_month, SUM([Total Excluding Tax]) AS sales, SUM([Profit]) AS profit FROM Fact.Sale GROUP BY DATEPART(year,[Invoice Date Key]), DATEPART(month,[Invoice Date Key]) ORDER BY sales_year, sales_month"""
with closing(connect()) as connection:
    monthly_sales = pd.read_sql_query(sql, connection)
monthly_sales
```
